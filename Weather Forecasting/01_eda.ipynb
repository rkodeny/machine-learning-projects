## Weather Forecasting - 01 Exploratory Data Analysis

A time-series project that forecats **hourly temperature** at Jena, Germany from its own recent history (lags) ans calender features. We explore the series' trend, seasonality, and autocorrelation - the structure the forecasting models will exploit in 03 model building.

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import warnings 
warnings.filterwarnings('ignore')
import utils

df = utils.load_data()
print('rows:', len(df), '| from', df.date.min(), 'to', df.date.max())
df.head()
```

### 1. Series over time

```python
fig, ax = plt.subplots(figsize=(13, 3.5))
ax.plot(df['date'], df['y'], lw=0.7, color="#4c72b0")
ax.set_ylabel("Temperature (°C)")
ax.set_title("Weather Forecasting : Temperature Over Time")
plt.tight_layout()
plt.show()
```

### 2. Distribution & Summary

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.hist(df['y'], bins=50, color='#55a868')
ax.set_title("Value Distribution")
plt.tight_layout()
plt.show()
print(df['y'].describe().round(2).to_string())
```

### 3. Trend(rolling mean)

```python
w = utils.SEASONAL_PERIOD * 7
fig, ax = plt.subplots(figsize=(13, 3.5))
ax.plot(df['date'], df['y'], lw = 0.5, alpha=0.4, label='actual')
ax.plot(df['date'], df['y'].rolling(w,min_periods=1).mean(), color="#c44e52", lw=2, label=f'rolling mean ({w})')
ax.legend()
ax.set_title('Trend via rolling mean')
plt.tight_layout()
plt.show()
```

### 4. Seasonal Decomposition (STL)

STL splits the series into **trend + season + residual** components - exposing a very strong daily cycle(warm afternoons, cold nights)
plus an anaual season.

```python
from statsmodels.tsa.seasonal import STL
s = df.set_index('date')['y'].asfreq(utils.FREQ).interpolate()
res= STL(s, period=utils.SEASONAL_PERIOD, robust=True).fit()
fig = res.plot()
fig.set_size_inches(12,7)
plt.tight_layout()
plt.show()
```

### 5. Season Profile(by hour)

```python
prof = df.assign(hour=df['date'].dt.hour).groupby('hour')['y'].mean()
fig, ax = plt.subplots(figsize=(8,3))
prof.plot(kind='bar', ax=ax, color="#8172b3")
ax.set_title("Mean Teamperature by hour")
plt.tight_layout()
plt.show()
```

### 6. Autocorrelation(ACF)
How strongly each value correlates with its past - the spikes shows which **lags** are informative features.

```python
from statsmodels.graphics.tsaplots import plot_acf
fig, ax = plt.subplots(figsize=(10, 3))
plot_acf(df['y'].dropna(), lags=min(40, len(df)//2-1), ax=ax)
plt.tight_layout()
plt.show()
```

### 7. Summary
>The series shows a very strong daily cycle (warm afternoons, cool nights) plus an annual season.  
Strong autocorrelation at short lags and at the daily (24-hour) period justifies **lag + rolling + calendar** features.

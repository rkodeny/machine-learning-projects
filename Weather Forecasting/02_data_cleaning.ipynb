## Weather Forecasting — 02 Feature Engineering & Split
Time series can't be shuffled. We build **lag features** (past values), **rolling mean/std**, and **calender** features, then split **chronologically** (train on the past, test on the most recent slice).

```python
import pandas as pd
import utils

df = utils.load_data()
print('raw rows:', len(df))
```

### 1. Build Features(lags + rolling + calender)

```python
feat=utils.build_features(df)
print('featured rows:', len(feat), '| features:', len(utils.feature_columns(feat)))
print('featured columns:', utils.feature_columns(feat))
feat.head()
```

### 2. Chronological Train/Test Split 

```python
train, test = utils.chronological_split(feat, test_frac=0.2)
print('train:', len(train), '(', train.date.min().date(),'->',train.date.max().date(),')')
print('test:', len(test), '(', test.date.min().date(),'->',test.date.max().date(),')')
```

### 3. Save featured data

```python
feat.to_csv('featured.csv', index=True)
print('saved featured.csv', feat.shape)
```
